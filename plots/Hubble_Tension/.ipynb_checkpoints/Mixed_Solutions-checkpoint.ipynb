# Geometric requirements for mixed solutions to the Hubble tension

This notebook produces a geometric visualization of the fixed-acoustic-scale condition

$$\delta\ln H_0=\delta\ln I-\delta\ln r_s,\qquad I\equiv\int_0^{z_*}\frac{dz}{E(z)}.$$

The figure illustrates how early-time modifications of the sound horizon and late-time modifications of the dimensionless expansion history must combine to increase (H_0) while preserving the CMB acoustic scale. Purely early- and late-time solutions correspond to motion along the horizontal and vertical axes, respectively, whereas mixed solutions populate the intermediate portion of the constant-(H_0) contour. An earlier version of this visualization was presented in a [CosmoVerse talk](https://www.youtube.com/watch?v=3Vmj9mLkmtE).

```python
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

def plot_rs_I_plane_clean(
    lim=0.10,
    H0_cmb=67.4,
    H0_sh0es=73.0,
    save=None,
):
    # symmetric plane around (0,0)
    xmin, xmax = -lim, lim
    ymin, ymax = -lim, lim

    # target shift needed to go from CMB to SH0ES
    dlnH0_target = np.log(H0_sh0es / H0_cmb)

    xx = np.linspace(xmin, xmax, 500)
    yy = np.linspace(ymin, ymax, 500)
    X, Y = np.meshgrid(xx, yy)

    # imposing delta ln theta_* = 0:
    # delta ln H0 = delta ln(integral) - delta ln r_s
    dlnH0 = Y - X

    fig, ax = plt.subplots(figsize=(8, 8))

    # centered diverging colormap
    vabs = np.max(np.abs(dlnH0))
    norm = mpl.colors.TwoSlopeNorm(vmin=-vabs, vcenter=0.0, vmax=vabs)

    cf = ax.contourf(
        X, Y, dlnH0,
        levels=200,
        cmap='coolwarm',
        norm=norm
    )

    # zero contour: delta ln H0 = 0  -> y = x
    ax.contour(
        X, Y, dlnH0,
        levels=[0.0],
        colors='black',
        linewidths=2.2
    )

    # target contour: delta ln H0 = dlnH0_target  -> y = x + dlnH0_target
    ax.contour(
        X, Y, dlnH0,
        levels=[dlnH0_target],
        colors='black',
        linewidths=2.2,
        linestyles='--'
    )

    # axes through origin
    ax.axhline(0.0, color='black', lw=1.1, alpha=0.8)
    ax.axvline(0.0, color='black', lw=1.1, alpha=0.8)

    # labels in the plane
    ax.text(-0.075, 0.015,
            r'$H_0=73$ km/s/Mpc',
            fontsize=13, rotation=45, color='black')

    # labels in the plane
    ax.text(0.03, 0.04,
            r'$H_0=67.4$ km/s/Mpc',
            fontsize=13, rotation=45, color='black')

    # LCDM point
    ax.scatter(0.0, 0.0, s=85, color='black', zorder=5)
    ax.text(0.004, -0.008, r'$\Lambda$CDM', fontsize=14, color='black')

    #Ealry
    ax.text(-0.069, 0.001,
            r'Early-time only',
            fontsize=14, rotation=0, color='darkred')

    #Late
    ax.text(-0.007, 0.015,
            r'Late-time only',
            fontsize=14, rotation=90, color='Darkgreen')


    # late-time-only arrow: delta ln r_s = 0, so move along x=0
    ax.annotate(
        "",
        xy=(0.0, dlnH0_target),
        xytext=(0.0, 0.0),
        arrowprops=dict(arrowstyle="-|>", lw=2.4, color="darkgreen"),
        zorder=6
    )

    # early-time-only arrow: schematic direction y=0
    ax.annotate(
        "",
        xy=(-dlnH0_target, 0.0),
        xytext=(0.0, 0.0),
        arrowprops=dict(arrowstyle="-|>", lw=2.4, color="darkred"),
        zorder=6
    )

    # axes labels
    ax.set_xlabel(r'$\delta \ln r_s$', fontsize=18)
    ax.set_ylabel(r'$\delta \ln \left(\int_0^{z_*}\frac{dz}{E(z)}\right)$', fontsize=18)

    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    ax.set_aspect('equal', adjustable='box')
    ax.tick_params(labelsize=13)

    # ticks
    ticks = [-0.09, -0.06, -0.03, 0.00, 0.03, 0.06, 0.09]
    ax.set_xticks(ticks)
    ax.set_yticks(ticks)

    # compact colorbar
    cbar = plt.colorbar(cf, ax=ax, pad=0.015, shrink=0.72, fraction=0.045, aspect=22)
    cbar.set_label(r'$\delta \ln H_0$', fontsize=17)
    cbar.ax.tick_params(labelsize=12)

    for spine in ax.spines.values():
        spine.set_linewidth(1.1)

    plt.tight_layout()

        # hatch region below the LCDM-H0 diagonal: y < x
    mpl.rcParams['hatch.linewidth'] = 0.15
    ax.contourf(

        X, Y, dlnH0,

        levels=[dlnH0.min(), 0.0],

        colors='none',

        hatches=['X'],

        zorder=3

    )

    if save is not None:
        plt.savefig(save, dpi=220, bbox_inches='tight')

    plt.show()

plot_rs_I_plane_clean(save="mixed.pdf")
```
